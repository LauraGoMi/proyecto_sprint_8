*Verificar y corregir tipos de datos (si aplica)*


Es importante verificar que cada columna tenga el tipo de dato correcto:
 - Texto → 'object'

 - Números decimales → 'float64'
 -  Fechas → 'datetime64'

Para el Dataset 1:
 
 - company_name debe ser 'object' (texto)
 - trips_amount debe ser 'int64' o 'float64' (número)

Para el Dataset 2:

 - dropoff_location_name debe ser 'object' (texto)

```python

print("🔧 VERIFICACIÓN DE TIPOS DE DATOS")
print("-" * 80)

# ----------------------
# Dataset 1
# ----------------------
print("\nTipos de datos - Dataset 1:")
print(trips_by_company.dtypes)

# Si trips_amount no es numérico, lo convertimos:
if trips_by_company['trips_amount'].dtype == 'object':
    print("\n⚠️  trips_amount es texto, convirtiendo a numérico...")
    trips_by_company['trips_amount'] = pd.to_numeric(
        trips_by_company['trips_amount'], 
        errors='coerce'  # Convierte valores inválidos en NaN
    )
    print("✅ Conversión completada")
else:
    print("✅ trips_amount ya es numérico")


# Dataset 2

print("\nTipos de datos - Dataset 2:")
print(dropoff_by_neighborhood.dtypes)


# Si average_trips no es numérico, lo convertimos:
if dropoff_by_neighborhood['average_trips'].dtype == 'object':
    print("\n⚠️  average_trips es texto, convirtiendo a numérico...")
    dropoff_by_neighborhood['average_trips'] = pd.to_numeric(
        dropoff_by_neighborhood['average_trips'], 
        errors='coerce'
    )
    print("✅ Conversión completada")
else:
    print("✅ average_trips ya es numérico")

print()
```

*Creacion de visualizaciones*

```python

print("📊  CREACIÓN DE VISUALIZACIONES")
print("-" * 80)

# ----------------------
# GRÁFICO 1: Top 10 compañías por viajes
# ----------------------
print("\n📈 Creando gráfico 1: Top 10 empresas de taxis por número de viajes...")

#  Seleccionar SOLO las top 10 compañías con más viajes

top_10_companies = trips_by_company.sort_values(
    by='trips_amount', 
    ascending=False  # De mayor a menor
).head(10)  # Tomamos solo las primeras 10

# Confirmamos que tenemos exactamente 10 compañías
print(f"   Graficando exactamente {len(top_10_companies)} compañías")

# Reordenamos para el gráfico (menor a mayor para que el #1 quede arriba)
top_10_companies_sorted = top_10_companies.sort_values(
    by='trips_amount', 
    ascending=True  
)

# Creamos una figura (ventana) para el gráfico

plt.figure(figsize=(14, 8))

#  Creamos un gráfico de barras horizontal SOLO con las top 10
plt.barh(
    y=top_10_companies_sorted['company_name'], 
    width=top_10_companies_sorted['trips_amount'],
    color='skyblue',
    edgecolor='navy',
    alpha=0.7  # Transparencia (0=transparente, 1=opaco)
)

# Añadimos título y etiquetas
plt.title(
    'Top 10 Compañías de Taxis por Número de Viajes (15-16 Nov 2017)', 
    fontsize=16, 
    fontweight='bold',
    pad=20  # Espacio entre título y gráfico
)
plt.xlabel('Número de Viajes', fontsize=12, fontweight='bold')
plt.ylabel('Compañía', fontsize=12, fontweight='bold')

#  Añadimos valores numéricos al lado de cada barra

for i, (name, value) in enumerate(zip(top_10_companies_sorted['company_name'], 
                                        top_10_companies_sorted['trips_amount'])):
    plt.text(
        value + 50,      # Posición x (un poco a la derecha de la barra)
        i,               # Posición y (altura de la barra)
        f'{value:,}',    # Texto a mostrar (valor con formato de miles)
        va='center',     # Alineación vertical centrada
        fontsize=10,
        fontweight='bold'
    )

plt.grid(axis='x', alpha=0.3, linestyle='--')
plt.tight_layout()

# Guardamos el gráfico como imagen
plt.savefig('grafico_top10_companias.png', dpi=300, bbox_inches='tight')
print("✅ Gráfico guardado como 'grafico_top10_companias.png'")

plt.show()

# ----------------------
# GRÁFICO 2: Top 10 barrios 
# ----------------------
print("\n📈 Creando gráfico 2: Top 10 barrios por finalizaciones...")

plt.figure(figsize=(14, 8))

# IMPORTANTE: Ya tenemos top_10_neighborhoods del Paso 5

top_10_sorted = top_10_neighborhoods.sort_values(
    by='average_trips', 
    ascending=True  
)

#  Confirmamos que tenemos exactamente 10 barrios
print(f"   Graficando exactamente {len(top_10_sorted)} barrios")

# Creamos un gráfico de barras horizontal SOLO con estos 10 barrios
plt.barh(
    y=top_10_sorted['dropoff_location_name'],  # Solo los 10 nombres
    width=top_10_sorted['average_trips'],      # Solo los 10 valores
    color='coral',
    edgecolor='darkred',
    alpha=0.7
)
# título y etiquetas
plt.title(
    'Top 10 Barrios por Número de Finalizaciones (Nov 2017)', 
    fontsize=16, 
    fontweight='bold',
    pad=20
)
plt.xlabel('Promedio de Viajes', fontsize=12, fontweight='bold')
plt.ylabel('Barrio', fontsize=12, fontweight='bold')


for i, (name, value) in enumerate(zip(top_10_sorted['dropoff_location_name'], 
                                        top_10_sorted['average_trips'])):
    plt.text(
        value + 5,  # Posición x (un poco a la derecha de la barra)
        i,          # Posición y (altura de la barra)
        f'{value:.1f}',  # Texto a mostrar (valor con 1 decimal)
        va='center',     # Alineación vertical centrada
        fontsize=10,
        fontweight='bold'
    )


plt.grid(axis='x', alpha=0.3, linestyle='--')


plt.tight_layout()


plt.savefig('grafico_top10_barrios.png', dpi=300, bbox_inches='tight')
print("✅ Gráfico guardado como 'grafico_top10_barrios.png'")


plt.show()

print()

```

*Preparacion de Datos*
- La columna start_ts aparece como Object, debemos convertida a Date
- Verificamos que duration_seconds sea numérico
- Eliminar valores faltantes si los hay
- Eliminar valores atípicos extremos, Por ejemplo, viajes de duración 0 o negativos

```python
print("🔧 PREPARACIÓN DE DATOS")
print("-" * 80)

# Convertir start_ts a datetime si es necesario
if trips_saturday['start_ts'].dtype == 'object':
    trips_saturday['start_ts'] = pd.to_datetime(trips_saturday['start_ts'])
    print("✅ Columna start_ts convertida a datetime")

# Verificar que duration_seconds sea numérico
if trips_saturday['duration_seconds'].dtype == 'object':
    trips_saturday['duration_seconds'] = pd.to_numeric(
        trips_saturday['duration_seconds'], 
        errors='coerce'
    )
    print("✅ Columna duration_seconds convertida a numérico")

# Eliminar valores faltantes si los hay
if trips_saturday.isnull().sum().sum() > 0:
    print(f"\n⚠️  Eliminando {trips_saturday.isnull().sum().sum()} valores faltantes...")
    trips_saturday = trips_saturday.dropna()
    print("✅ Valores faltantes eliminados")

# Eliminar valores atípicos extremos 

print(f"\nViajes con duración <= 0: {(trips_saturday['duration_seconds'] <= 0).sum()}")
trips_saturday = trips_saturday[trips_saturday['duration_seconds'] > 0]

print(f"\n✅ Dataset limpio: {trips_saturday.shape[0]} viajes")
print()
```

*Separar Datos por condicion climaticas*

-Separamos los viajes en dos grupos según el clima:

Grupo 1: Sábados con mal clima (Bad)

Grupo 2: Sábados con buen clima (Good)

*Calcular estadisticas descriptivas del Grupo 1 y Grupo 2*

*Formulacion de Hipotesis*

HIPÓTESIS DEL ESTUDIO:

"La duración promedio de los viajes desde el Loop hasta el Aeropuerto  Internacional O'Hare CAMBIA los sábados lluviosos"


H₀ (Hipótesis Nula):
   "NO hay diferencia en la duración promedio de viajes entre sábados 
   lluviosos y sábados no lluviosos"
   

H₁ (Hipótesis Alternativa):
   "SÍ hay diferencia en la duración promedio de viajes entre sábados 
   lluviosos y sábados no lluviosos"
   

**TIPO DE TEST**:
   Test de dos colas (two-tailed test), Porque buscamos cualquier DIFERENCIA, no una dirección específica

TEST ELEGIDO: Prueba t de Student para muestras independientes
              (Independent Samples t-test)

1. ✅ COMPARAMOS DOS GRUPOS INDEPENDIENTES:
   
   • Grupo 1: Viajes en sábados con mal clima
   
   • Grupo 2: Viajes en sábados con buen clima
   
   • Son independientes porque un viaje no puede estar en ambos grupos



 Elegimos alpha = 0.05, que es el mas recomendado para éste tipo de analisis ya que es el valor estándar en ciencias sociales y análisis de negocios
 


```python

print("🧮 EJECUCIÓN DEL TEST ESTADÍSTICO")
print("=" * 80)

# Realizamos el test t para muestras independientes
# stats.ttest_ind() realiza la prueba t
# Devuelve dos valores:
# - t_statistic: el valor del estadístico t
# - p_value: la probabilidad (valor p)

alpha = 0.05


t_statistic, p_value = stats.ttest_ind(
    bad_weather_trips,    # Grupo 1: mal clima
    good_weather_trips,   # Grupo 2: buen clima
    equal_var=False       # No asumimos varianzas iguales (Welch's t-test)
)

# equal_var=False significa que usamos la corrección de Welch
# Esto es más seguro cuando no sabemos si las varianzas son iguales

print("\n📊 RESULTADOS DEL TEST:")
print("-" * 80)
print(f"\nEstadístico t: {t_statistic:.4f}")
print(f"Valor p (p-value): {p_value:.6f}")
print(f"Nivel de significancia (α): {alpha}")

# Interpretación del estadístico t:
# - Si t es grande (en valor absoluto), hay mayor diferencia entre grupos
# - Si t está cerca de 0, los grupos son similares

print(f"\n📈 INTERPRETACIÓN DEL ESTADÍSTICO t:")
print(f"   • t = {t_statistic:.4f}")
if abs(t_statistic) > 2:
    print(f"   • |t| > 2 → Diferencia considerable entre grupos")
else:
    print(f"   • |t| ≤ 2 → Diferencia pequeña entre grupos")

print()

```

<div class="alert alert-block alert-warning"> 
<b>Comentario del Revisor #1</b> <a class="tocSkip"></a>

Al realizar una prueba de hipótesis, es muy importante verificar los supuestos. El t-test que usaste funciona bien cuando se cumplen ciertos supuestos, como que las varianzas de ambos grupos (sábados lluviosos y no lluviosos) sean similares. Puedes realizar la prueba de levene para comprobar que los dos grupos tienen la misma varianza y, de esta forma, podrás saber el valor del parámetro equal_var que se debe utilizar en la función stats.ttest_ind.

*Decision*

```python

print("🎯 DECISIÓN Y CONCLUSIÓN")
print("=" * 80)

print(f"\nCOMPARACIÓN:")
print(f"   • p-value = {p_value:.6f}")
print(f"   • α = {alpha}")
print(f"   • p-value {'<' if p_value < alpha else '≥'} α")

print("\n" + "=" * 80)

if p_value < alpha:
    print("✅ RECHAZAMOS la hipótesis nula (H₀)")
    print("=" * 80)
   
```
